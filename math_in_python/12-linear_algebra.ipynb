```python
# row and clolumn vectors
# vector -- ordered list of numbers
# dimensionality -- number of components in a vector

import numpy as np
import matplotlib.pyplot as plt
import sympy as sym
from IPython.display import display, Math
```

```python
# row vector
row_vec = np.array([4, 2])
col_vec = np.array([ [-2], [3] ])

print("row vector:", row_vec)
print("column vector:\n", col_vec)
```

```python
display(Math(sym.latex(sym.simplify(row_vec))))
display(Math(sym.latex(sym.simplify(col_vec))))
```

```python
plt.plot([0, row_vec[0]], [0, row_vec[1]], 'r', label='row vector')
plt.plot([0, col_vec[0][0]], [0, col_vec[1][0]], 'b', label='column vector')

plt.axis('square')
plt.axis([-5, 5, -5, 5])
plt.grid()
plt.legend()
plt.show()
```

```python
row_vec_1 = np.array([3, 0, -4])
row_vec_2 = np.array([-1, 1, 3])
```

```python
# for 3D plotting
from mpl_toolkits.mplot3d import Axes3D
```

```python
fig = plt.figure(figsize = plt.figaspect(1))
ax = fig.add_subplot(1, 1, 1, projection='3d')

ax.plot([0, row_vec_1[0]], [0, row_vec_1[1]], [0, row_vec_1[2]], 'r', label='row vector 1')
ax.plot([0, row_vec_2[0]], [0, row_vec_2[1]], [0, row_vec_2[2]], 'b', label='row vector 2')
ax.legend()

ax.plot([-5, 5], [0, 0], [0, 0], '--', color='gray')
ax.plot([0, 0], [-5, 5], [0, 0], '--', color='gray')
ax.plot([0, 0], [0, 0], [-5, 5], '--', color='gray')

ax.set_xlim([-5, 5])
ax.set_ylim([-5, 5])
ax.set_zlim([-5, 5])
plt.show()
```

```python
v = np.array([.5, 1])
s = [1, -.5, 2, .5]

for si in s:
    sv = si * v
    plt.plot([0, sv[0]], [0, sv[1]], 'o-', linewidth=3, label=f'$\\lambda$ = {si}')

plt.axis('square')
plt.axis([-3, 3, -3, 3])
plt.grid()
plt.legend()
plt.show()
```

```python
v1 = np.array([-1, 2])
v2 = np.array([1, 1])

v3a = v1 + v2

v3b = np.add(v1, v2)

v3c = np.zeros(2)
for i in range(0, 2):
    v3c[i] = v1[i] + v2[i]

print(v3a, v3b, v3c)
```

```python
plt.plot([0, v1[0]], [0, v1[1]], 'r', label='v1')
plt.plot([v1[0], v3a[0]], [v1[1], v3a[1]], 'b', label='v2')
plt.plot([0, v3a[0]], [0, v3a[1]], 'g', label='v1 + v2')
plt.axis('square')
plt.axis([-3, 3, -3, 3])
plt.grid()
plt.legend()
plt.show()
```
