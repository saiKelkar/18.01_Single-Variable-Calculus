```python
# row and clolumn vectors
# vector -- ordered list of numbers
# dimensionality -- number of components in a vector

import numpy as np
import matplotlib.pyplot as plt
import sympy as sym
from IPython.display import display, Math
```

```python
# row vector
row_vec = np.array([4, 2])
col_vec = np.array([ [-2], [3] ])

print("row vector:", row_vec)
print("column vector:\n", col_vec)
```

```python
display(Math(sym.latex(sym.simplify(row_vec))))
display(Math(sym.latex(sym.simplify(col_vec))))
```

```python
plt.plot([0, row_vec[0]], [0, row_vec[1]], 'r', label='row vector')
plt.plot([0, col_vec[0][0]], [0, col_vec[1][0]], 'b', label='column vector')

plt.axis('square')
plt.axis([-5, 5, -5, 5])
plt.grid()
plt.legend()
plt.show()
```
