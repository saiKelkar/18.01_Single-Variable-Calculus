```python
import numpy as np
import sympy as sym
from IPython.display import display, Math
```

```python
x = sym.symbols('x')

expr = 2 * x + 4 - 9
sym.solve(expr)
```

```python
display(Math(f'\\text{{The solution to }} {sym.latex(expr)} \\text{{ is }} {sym.latex(sym.solve(expr)[0])}'))
```

```python
expr = x**2 - 4 
sol = sym.solve(expr)

for i in range(len(sol)):
    display(Math(f'\\text{{The solution to }} {sym.latex(expr)} \\text{{ is }} {sym.latex(sol[i])}'))
```

```python
y = sym.symbols('y')

expr = x / 4 - x * y + 5
sym.solve(expr, x)

display(Math(f'\\text{{The solution to }} {sym.latex(expr)} \\text{{ for }} x \\text{{ is }} {sym.latex(sym.solve(expr, x)[0])}'))
```

```python
q = sym.symbols('q')

expr = 3 * q + 4 / q + 3 - 5 * q - 1 / q - 1
display(Math(f'\\text{{The solution to }} {sym.latex(expr)} \\text{{ for }} q \\text{{ is }} {sym.latex(sym.solve(expr, q))}'))
```
