```python
import numpy as np
import sympy as sym
from IPython.display import display, Math
```

```python
x = sym.symbols('x')

expr = 2 * x + 4 - 9
sym.solve(expr)
```

```python
display(Math(f'\\text{{The solution to }} {sym.latex(expr)} \\text{{ is }} {sym.latex(sym.solve(expr)[0])}'))
```

```python
expr = x**2 - 4 
sol = sym.solve(expr)

for i in range(len(sol)):
    display(Math(f'\\text{{The solution to }} {sym.latex(expr)} \\text{{ is }} {sym.latex(sol[i])}'))
```

```python
y = sym.symbols('y')

expr = x / 4 - x * y + 5
sym.solve(expr, x)

display(Math(f'\\text{{The solution to }} {sym.latex(expr)} \\text{{ for }} x \\text{{ is }} {sym.latex(sym.solve(expr, x)[0])}'))
```

```python
q = sym.symbols('q')

expr = 3 * q + 4 / q + 3 - 5 * q - 1 / q - 1
display(Math(f'\\text{{The solution to }} {sym.latex(expr)} \\text{{ for }} q \\text{{ is }} {sym.latex(sym.solve(expr, q))}'))
```

```python
from sympy.abc import x

term_1 = (4 * x + 5)
term_2 = x

print(term_1 * term_2)
print(sym.expand(term_1 * term_2))
display(Math(sym.latex(sym.expand(term_1 * term_2))))
```

```python
term_3 = x - 7

display(Math(sym.latex(term_1 * term_3)))
display(Math(sym.latex(sym.expand(term_1 * term_3))))
```

```python
from sympy.abc import y 

expr = x * (2 * y ** 2 - 5 ** x / x)
display(Math(sym.latex(expr)))
display(Math(sym.latex(sym.expand(expr))))
```

```python
expr = (4 + x) * (2 - y)
display(Math(sym.latex(expr)))
display(Math(sym.latex(sym.expand(expr))))
```

```python
for i in range(3):
    for j in range(3):
        display(Math(f'{sym.latex(sym.expand(expr))} \\text{{ evaluated at }} x={i}, y={j} \\text{{ is }} {sym.latex(expr.subs({x: i, y: j}))}'))
```

```python
# associative
from sympy.abc import x, y, z

expr1 = x * (4 * y)
expr2 = (x * 4) * y

expr1 == expr2
```

```python
# commutative

e1 = 4 * x * y
e2 = x * 4 * y
e3 = y * x * 4
```

```python
e1.subs({x: 3, y: 2})
```

```python
# distributive
a, b, c, d = sym.symbols('a b c d')

expr = (a + b) * (c + d)
sym.expand(expr)
```
